```python
import logging
import requests
import time
import os
```

```python
os.makedirs("../logs", exist_ok=True)
```

```python
# Logging Configuration #
logging.basicConfig(
    filename="../logs/weather.log",
    level=logging.INFO,
    format="%(asctime)s | %(levelname)s | %(message)s"
)
logger = logging.getLogger("WeatherLogger")
```

```python
# function to fetch weather #
def get_weather(latitude, longitude):
  url = (
      f"https://api.open-meteo.com/v1/forecast?"
        f"latitude={latitude}&longitude={longitude}&current_weather=true"
  )
  logger.info(f"Sending request to Weather API: {url}")
  start_time = time.time()
  try:
        response = requests.get(url, timeout=10)
        elapsed = time.time() - start_time

        # Slow response warning (>1.5 sec)
        if elapsed > 1.5:
            logger.warning(f"Slow response detected: {elapsed:.2f} seconds")

        if response.status_code == 200:
            logger.info("Weather API response received successfully")
            data = response.json()
            return data["current_weather"]
        else:
            logger.error(f"API returned error: {response.status_code}")
            return None

  except Exception as e:
        logger.error(f"Exception during API call: {e}")
        return None
```

```python
if __name__ == "__main__":
  print("Weather Logger Started!\nCheck logs/weather.log for entries.")
  latitude = 18.5204
  longitude = 73.8567
  weather = get_weather(latitude, longitude)
  if weather:
    print("\nCurrent Weather:")
    print("-----------------")
    print(f"Teamperature: {weather['temperature']}*C")
    print(f"Windspeed: {weather['windspeed']}kn/h")
    print(f"Wind Direction: {weather['winddirection']}*")
    print(f"Weather Code: {weather['weathercode']}")
    logger.info(f"Weather Data: {weather}")
  else:
    print("Failed to fetch weather data.")
```
