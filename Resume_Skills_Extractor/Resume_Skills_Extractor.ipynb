```python
import PyPDF2
import nltk
from nltk import word_tokenize, ne_chunk
from nltk.tag import PerceptronTagger

# initialize tagger
tagger = PerceptronTagger()

# ---------------------------------------
#   1. Extract TEXT from PDF
# ---------------------------------------
def extract_text_from_pdf(pdf_path):
    text = ""
    try:
        with open(pdf_path, "rb") as f:
            reader = PyPDF2.PdfReader(f)
            for page in reader.pages:
                text += page.extract_text() + "\n"
    except Exception as e:
        print("Error reading PDF:", e)
    return text


# ---------------------------------------
#   2. SKILLS Extraction
# ---------------------------------------
skills_keywords = [
    'python','java','c++','sql','nlp','machine learning','deep learning',
    'tensorflow','keras','pandas','numpy','statistics','git','linux',
    'power bi','data visualization','data analysis','excel','html','css'
]

def extract_skills(text):
    tokens = [t.lower() for t in word_tokenize(text)]
    found = set()

    for skill in skills_keywords:
        skill_tokens = skill.split()
        if len(skill_tokens) == 1:
            if skill in tokens:
                found.add(skill.title())
        else:
            if skill in text.lower():
                found.add(skill.title())
    return list(found)


# ---------------------------------------
#   3. EDUCATION Extraction
# ---------------------------------------
education_keywords = [
    'bachelor', 'master', 'b.tech', 'm.tech', 'b.e', 'phd',
    'computer science', 'engineering', 'degree', 'university', 'college'
]

def extract_education(text):
    sentences = nltk.sent_tokenize(text)
    edu = []
    for sent in sentences:
        for kw in education_keywords:
            if kw in sent.lower():
                edu.append(sent)
                break
    return edu


# ---------------------------------------
#   4. EXPERIENCE Extraction
# ---------------------------------------
def extract_experience(text):
    sentences = nltk.sent_tokenize(text)
    experience_list = []
    keywords = ["experience", "intern", "developer", "engineer", "worked", "project"]

    for sent in sentences:
        if any(k in sent.lower() for k in keywords):
            experience_list.append(sent)
    return experience_list


# ---------------------------------------
#   5. Named Entity Recognition (NER)
# ---------------------------------------
def extract_ner(text):
    tokens = word_tokenize(text)
    pos = tagger.tag(tokens)   # using new tagger
    chunks = ne_chunk(pos)

    entities = []
    for chunk in chunks:
        if hasattr(chunk, 'label'):
            entity = " ".join(c[0] for c in chunk)
            entities.append((entity, chunk.label()))
    return entities


# ---------------------------------------
#   6. MAIN PIPELINE
# ---------------------------------------
def analyze_resume(pdf_path):
    text = extract_text_from_pdf(pdf_path)

    result = {
        "Skills": extract_skills(text),
        "Education": extract_education(text),
        "Experience": extract_experience(text),
        "Named Entities": extract_ner(text)
    }
    return result
```
